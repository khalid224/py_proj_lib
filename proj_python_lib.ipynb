```python
import requests
import pandas as pd

url = "https://dummyjson.com/users?limit=100"

response = requests.get(url)
data = response.json()


df = pd.json_normalize(data["users"])

df.head()
```

```python
df["age"].fillna(df["age"].median(), inplace=True)
df["height"].fillna(df["height"].median(), inplace=True)
df["weight"].fillna(df["weight"].median(), inplace=True)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
sns.displot(data=df, x="age", kde=True)
```

```python
sns.displot(data=df, x="role", kde=True)
```

```python
sns.displot(data=df, x="age",hue="role" ,kde=True)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="gender", data=df)
plt.title("Number of Users per Gender")
plt.show()
```

```python
df.rename(columns={"address.city": "city"}, inplace=True)
top_cities = df["city"].value_counts().head(10)
top_cities
```

```python
plt.figure(figsize=(8,5))
sns.barplot(
    y=top_cities.index,
    x=top_cities.values
)
plt.title("Top 10 Cities by Number of Users")
plt.xlabel("Number of Users")
plt.ylabel("City")
plt.show()
```

```python
avg_height = df["height"].mean()
avg_weight = df["weight"].mean()

```

```python
plt.figure(figsize=(12,4))

plt.subplot(1,2,1)
sns.histplot(df["height"], kde=True)
plt.title(f"Height Distribution (Mean = {avg_height:.1f} cm)")

plt.subplot(1,2,2)
sns.histplot(df["weight"], kde=True)
plt.title(f"Weight Distribution (Mean = {avg_weight:.1f} kg)")

plt.tight_layout()
plt.show()
```
